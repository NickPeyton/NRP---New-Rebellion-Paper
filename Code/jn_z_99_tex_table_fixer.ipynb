# Quick Script to Fix the Tex Tables

```python
import os
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
RAW = PROJECT_ROOT / 'Data' / 'Raw'
PROCESSED = PROJECT_ROOT / 'Data' / 'Processed'
TABLES = PROJECT_ROOT / 'Output' / 'Tables'
```

```python
for file in os.listdir(TABLES):
    if not file.endswith('.tex'):
        continue
    with open(TABLES / file, 'r+') as f:
        content = f.read()
        if '\\begin{threeparttable}' in content:
            continue
        if '\\begin{table}[H] \centering' not in content:
            print('No begin table found in', file)
            continue
        content = content.replace('\\begin{table}[H] \centering', '\\begin{table}[H] \n\centering \nsinglespacing \n\\begin{threeparttable}')  
        if '\\hline \\\\[-1.8ex] \n\\textit{Note:}  & \\multicolumn{3}{r}{$^{*}$p$<$0.1; $^{**}$p$<$0.05; $^{***}$p$<$0.01} \\\\' not in content:
            print('No note found in', file)
            continue
        content = content.replace('\\hline \\\\[-1.8ex] \n\\textit{Note:}  & \\multicolumn{3}{r}{$^{*}$p$<$0.1; $^{**}$p$<$0.05; $^{***}$p$<$0.01} \\\\ \n', '')
        content = content.replace('\\end{tabular}', '\\end{tabular} \n\\begin{tablenotes} \n\\item * p$<$.1, ** p$<$.05, ***p$<$.01 \n\\footnotesize \n\\item FILL NOTE \n\end{tablenotes} \n\\end{threeparttable}')
        f.seek(0)
        f.write(content)
        f.truncate()
```
