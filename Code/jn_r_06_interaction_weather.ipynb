# Interaction Analysis: Monastic Land and Weather Shocks

This notebook tests whether the impact of monastic land on rebellion is amplified by adverse weather conditions (PDSI moisture shocks in 1535 and 1536).

Hypothesis: The loss of monastic services was more acute in parishes already suffering from harvest failure.

```r
pacman::p_load(
  sf, tidyverse, stargazer, dplyr,
  spdep, ggplot2, conleyreg, jsonlite
)

PROJECT_ROOT <- ".."
setwd(PROJECT_ROOT)
pretty_dict <- fromJSON("Code/pretty_dict.json")

pdf <- read_sf(dsn = "Data/Processed/northParishFlows.shp")

# Standardize continuous variables for easier interaction interpretation
continuous_vars <- c("lbg_arak", "lsm_arak", "loth_arak", "lown_arak", "wet_1535", "wet_1536", "lpopC", "lLStax_pc")
for (v in continuous_vars) {
  pdf[[v]] <- scale(pdf[[v]], center = TRUE, scale = TRUE)[, 1]
}

hide_vars <- c("area", "mean_slope", "uplands", "lowlands", "distScot", "Constant")
controls_formula <- "lpopC + lLStax_pc + area + mean_slope + uplands + lowlands + distScot"
```

## Section 1: Interaction with 1536 Weather (Harvest Year)

Testing `monastic_var * wet_1536`

```r
models_36 <- list(
  m1 = glm(paste0("primary ~ lbg_arak * wet_1536 + ", controls_formula), data = pdf, family = binomial(link = "logit")),
  m2 = glm(paste0("primary ~ loth_arak * wet_1536 + ", controls_formula), data = pdf, family = binomial(link = "logit")),
  m3 = glm(paste0("primary ~ lni_arak * wet_1536 + ", controls_formula), data = pdf, family = binomial(link = "logit"))
)

stargazer(models_36, type = "text", 
          column.labels = c("Large House", "Off-site", "Net Income"),
          omit = hide_vars)
```

## Section 2: Interaction with 1535 Weather (Pre-Rebellion Year)

Testing `monastic_var * wet_1535`

```r
models_35 <- list(
  m4 = glm(paste0("primary ~ lbg_arak * wet_1535 + ", controls_formula), data = pdf, family = binomial(link = "logit")),
  m5 = glm(paste0("primary ~ loth_arak * wet_1535 + ", controls_formula), data = pdf, family = binomial(link = "logit")),
  m6 = glm(paste0("primary ~ lni_arak * wet_1535 + ", controls_formula), data = pdf, family = binomial(link = "logit"))
)

stargazer(models_35, type = "text", 
          column.labels = c("Large House", "Off-site", "Net Income"),
          omit = hide_vars)
```
