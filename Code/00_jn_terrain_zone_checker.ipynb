```python
ndf = df[['T_ZONEGRP', 'T_ZONE', 'T_ZONEDSC', 'geometry']]
```

```python
# Group by Zone but still output group and desc so we can analyze each type of terrain zone
grouped = ndf.groupby('T_ZONE').agg({'T_ZONEGRP': 'first', 'T_ZONEDSC': 'first'}).reset_index()
```
