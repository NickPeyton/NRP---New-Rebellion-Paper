```python
import pandas as pd
import statsmodels.api as sm
```

```python
df = pd.read_csv(r"C:\Users\nicho\My Drive\20_Projects\NRP---New-Rebellion-Paper\Data\Raw\CSV\clark_land_vals.csv")
```

```python
# Set year > 1500 and year < 1600
df = df[(df['year'] > 1500) & (df['year'] < 1600)]

# Get _area and _value columns
area_cols = [col for col in df.columns if '_area' in col]
value_cols = [col for col in df.columns if '_value' in col]
print(area_cols)
print(value_cols)
```

```python
df['pastoral_area'] = df['pasture_area'] + df['meadow_area']
main_land_types = ['arable_area', 'pasture_area', 'meadow_area']
other_land_types = ['garden_area', 'orchard_area', 'wood_area', 'ozier_area', 'hops_area', 'fen_area', 'marsh_area']
buildings = ['houses', 'cottages', 'barns']

df['other_land_area'] = df[other_land_types].sum(axis=1)
```

```python
# Interceptless regression of annual rental value on arable, pastoral, and other land area
# Drop rows with missing main land types or annual rental value
df[buildings] = df[buildings].fillna(0)
df.dropna(subset=main_land_types + ['annual_rental_value'], inplace=True)
df['annual_rental_value'].dropna(inplace=True)
X = df[main_land_types + ['other_land_area'] + buildings]
y = df['annual_rental_value']
model = sm.OLS(y, X).fit()
print(model.summary())
```
